# Problem Statement: Find the best fit mode to purchasing car


### Importing Libraries:

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.linear_model import Lasso
from sklearn.linear_model import Ridge
```

```python
# Reading csv file 
df = pd.read_csv('car_purchasing.csv', encoding='ISO-8859-1')
```

# Data PreProcessing

## Consider only numerical data 

```python
df.drop(columns= ['customer name', 'customer e-mail', 'country', 'gender'], inplace= True)
```

```python
# Only newmerical Data
df.head()
```

```python
# Correlation
df.corr()
```

```python
# Figure it out x and y
x = df.iloc[:, : -1]
y = df.iloc[:, -1]
```

```python
y = y.values.reshape(-1, 1)
```

# Data Scalling 

```python
scaler = MinMaxScaler()
x = scaler.fit_transform(x)
y = scaler.fit_transform(y)
```

```python
x
```

```python
y
```

# Data Splitting

```python
len(x)
```

```python
len(y)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.1, shuffle=True, random_state=42)
```

```python
x_train.shape
```

```python
y_train.shape
```

```python
y_test.shape
```

# Machine Learning Model: Linear Regression

```python
lr_model =LinearRegression()
lr_model.fit(x_train, y_train)
```

```python
lr_model.score(x_train, y_train)
```

lr_pred = lr_model.predict(x_test)

```python
mean_squared_error(y_test, lr_pred)
```

```python
r2_score(y_test, lr_pred)
```

# Lasso Regression 

```python
lasso = Lasso(alpha = 0.0001)
lasso.fit(x_train, y_train)
```

```python
lasso.score(x_train, y_train)
```

```python
lasso_pred = lasso.predict(x_test)
```

```python
mean_squared_error(y_test, lasso_pred)
```

```python
r2_score(y_test, lasso_pred)
```

# Radge Regression

```python
ridge = Ridge(alpha=0.001)
ridge.fit(x_train, y_train)
```

```python
ridge.score(x_train, y_train)
```

```python
ridge_pred = ridge.predict(x_test)
```

```python
mean_squared_error(y_test, ridge_pred)
```

```python
r2_score(y_test, ridge_pred)
```

```python
plt.Figure(figsize=(10, 10))
plt.scatter(y_test, lr_pred, c='r', alpha=0.2, label = "Linear Regression")
plt.scatter(y_test, lasso_pred, c='g', alpha=0.2, label = "Linear Regression")
plt.scatter(y_test, ridge_pred, c='b', alpha=0.2, label = "Linear Regression")
plt.legend()
plt.show()
```
